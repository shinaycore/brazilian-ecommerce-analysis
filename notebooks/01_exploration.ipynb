```python
# Importing necessary libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
# Importing files into this notebook.
candidates = [
    Path("../data/raw"), Path("../data"),   # notebook running from notebooks/
    Path("data/raw"), Path("data"),         # notebook running from project root
]
DATA_DIR = next((p for p in candidates if (p / "olist_orders_dataset.csv").exists()), None)

if DATA_DIR is None:
    raise FileNotFoundError(
        f"Couldn't find the CSVs. Current folder: {Path.cwd()}\n"
        f"Checked: {[str(p) for p in candidates]}"
    )
print("Using data folder:", DATA_DIR.resolve())

files = {
    "customers":  "olist_customers_dataset.csv",
    "geo":        "olist_geolocation_dataset.csv",
    "items":      "olist_order_items_dataset.csv",
    "payments":   "olist_order_payments_dataset.csv",
    "reviews":    "olist_order_reviews_dataset.csv",
    "orders":     "olist_orders_dataset.csv",
    "products":   "olist_products_dataset.csv",
    "sellers":    "olist_sellers_dataset.csv",
    "category":   "product_category_name_translation.csv",
}

dfs = {name: pd.read_csv(DATA_DIR / f) for name, f in files.items()}
print({name: df.shape for name, df in dfs.items()})
```
